**Step 1 : Import pandas library**

```python
# Import pandas to work with data and tables.
# 'pd' is the short name used for pandas.
import pandas as pd
```

**Step 2 : create sample dataset**

```python
# Create employee details in a dictionary.
# It contains employee ID, name, department, city, salary, and experience.
data = {

    "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","PRIYA","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]

}
```

**Step 3 : Create DataFrame**

```python
# Convert the data into a DataFrame.
# A DataFrame stores data in rows and columns.
df=pd.DataFrame(data)
df
```

**Step 4  : Display the DataSet**

```python
# Display complete DataFrame
df
```

**Step 5:Filter the Data**

Task 1:Display Employee from IT Department

```python
#Select Employee working in IT Department
df[df["Department"]=="IT"]
```

Task 2 : Display Employee with salary Greater than 50,000

```python
#Display salary whose salary is greater than 50,000
df[df["Salary"]>50000]
```

Task 3 : Display Emplopyees from Surat

```python
# Select employees who live in Surat.
# Only records with Surat as the city are displayed.
df[df["City"]=="Surat"]
```

Task 4 : Apply Multiple Condition

```python
#Display Employee from IT dept
#Display salary greater than 50,000
df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

**Step 6 : sorting Data**

Task 1 : Sort salary in ascending order

```python
#sort employee by salary from lowest to highest
df.sort_values(by="Salary")
```

Task 2 : sort salary in desecending order

```python
# sort employee by salary in descending order
df.sort_values(by="Salary",ascending=False)
```

Task 3 : Sort Department and Salary

```python
#sort by Department
# within each Department sort Salaries in ascending order,Department iN DESCENDING ORDER
df.sort_values(by=["Department","Salary"],ascending =[True,False])


```

**Step 7 : Perform Aggregation**

Task 1 : find total salary

```python
# Add all employee salaries together.
# This gives the total salary of all employees
df["Salary"].sum()
```

Task 2: count average salary

```python
# Find the average salary of employees.
# This gives the mean salary of the dataset.
df["Salary"].mean()
```

Task 3 : display max salary

```python
# Find the highest salary in the dataset.
# max() returns the largest salary value.
df["Salary"].max()
```

Task 4 : Display min salary

```python
# Find the lowest salary in the dataset.
# min() returns the smallest salary value.
df["Salary"].min()
```

Task 5 : count total employees

```python
# Count the number of employees.
# The employee IDs are counted to get the total.
df["Emp_ID"].count()
```

**Step 8 :Perform GroupBY**

Task 1 : Display Average salary department_wise

```python
# Group employees according to their department.
# Find the average salary for each department.
df.groupby("Department")["Salary"].mean()
```

Task 2 : Display Minimum salary department-wise

```python
# Group employees by department.
# Find the lowest salary in each department.
df.groupby("Department")["Salary"].min()
```

Task 3 : count the Employees Department-Wise

```python
# Group employees by department.
# Count the number of employees in each department.
df.groupby("Department")["Emp_ID"].count()
```

Task 4 : Total Salary Department-Wise

```python
# Group employees according to their department.
# Find the total salary for each department.
df.groupby("Department")["Salary"].sum()
```

Task 5 : Display Maximum Salary Department-Wise

```python
# Group employees by department.
# Find the highest salary in each department.
df.groupby("Department")["Salary"].max()
```

Task 6 : Display City wise salary

```python
# Group employees according to their city.
# Find the average salary for each city.
df.groupby("City")["Salary"].mean()
```

**Step 9 :Multiple Aggregation Function**

Task : Multiple Aggregation by Department

```python
#Apply multiple statistical function
#to the Salary column for each Dapartment
df.groupby("Department")["Salary"].agg(["count","sum","mean","min","max"])
```

**Step 7: Save Result**

```python
#save the dataset into a CSV file
df.to_csv("Employee_Data.csv",index=False)
print("CSV file created successfully!")
```
