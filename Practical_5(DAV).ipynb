###**AIM : Analyze Dataset using Correlation,Covariance,Outlier detection and distribution analysis**

**Step 1: Import Require Libraries**

```python
#import require libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Step 2 : Create DataFrame**

```python
#create DataFrame using pandas
data = {
    'Student_ID':[101,102,103,104,105,106,107,108,109,110,111,112,113,114,115],
    'Study_Hours':[2,3,4,5,6,7,8,9,3,4,5,6,7,10,1],
    'Attandance':[65,70,75,80,82,85,90,92,68,74,78,84,88,95,40],
    'assignment_Score':[55,60,65,70,72,78,82,88,62,67,71,76,80,95,35],
    'Final_Marks':[50,55,62,68,72,78,82,90,58,63,69,75,81,98,25]
}
df=pd.DataFrame(data)
# Display dataframe
df
```

**Step 3 : Inspect Dataset**

```python
# Display First Five Records
df.head()
```

```python
# Display information about dataset including Not NULL values and Datatype
df.info()
```

```python
#Display no of rows and columns in Dataframe
df.shape
```

```python
#Generate basic descriptive statistics
df.describe()
```

**Step 4 : Calculate correlation**

```python
# Calculate correlation between Study_Hours and Final marks
Correlation = df['Study_Hours'].corr(df['Final_Marks'])
#Display Correlation
print("Correlation between Study_Hours and Final marks:", Correlation)
```

**Step 5 : Calculate Correlation matrix**

```python
#Calculate correlation among all numeric variables
Correlation_Matrix = df.corr()
#Display Correlation Matrix
Correlation_Matrix
```

**Step 6 : Visualize Correlation Using Scatter plot**

```python
#Create scatter plot to visualize the relation between Study_Hours and Final_Marks
plt.scatter(df['Study_Hours'], df['Final_Marks'])
plt.xlabel('Study_Hours')
plt.ylabel('Final_Marks')
plt.title('Scatter Plot of Study_Hours vs Final_Marks')
#Display plot
plt.show()
```

**Step 7 : Calculate covariance**

```python
#Calculate Covariance between Study_Hours and Final Marks
Covariance = df['Study_Hours'].cov(df['Final_Marks'])
#Display Covariance
print("Covariance between Study_Hours and Final marks:", Covariance)
```

**Step 8: Calculate Covariance Matrix**

```python
#Select numeric columns for covariance matrix
numeric_columns = df.select_dtypes(include=[np.number])
#Calculate Covariance Matrix
Covariance_Matrix = numeric_columns.cov()
#Display Covariance Matrix
Covariance_Matrix
```

**Step 9 : Detect Outliers using IQR**

```python
# Detect Outliers using IQR using Final_Marks column
Q1 = df['Final_Marks'].quantile(0.25)
Q2 = df['Final_Marks'].quantile(0.50)
Q3 = df['Final_Marks'].quantile(0.75)
IQR = Q3 - Q1
#Display Outliers
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

```

**Step 10 : Identify Outliers**

```python
#Identify the Outliers
Lower_Bound = Q1 - 1.5 * IQR
Upper_Bound = Q3 + 1.5 * IQR
print("Lower Bound:", Lower_Bound)
print("Upper Bound:", Upper_Bound)
#Select where Final_Marks are below the lower_bound or greater than Upper_bound
Outliers = df[(df['Final_Marks'] < Lower_Bound) | (df['Final_Marks'] > Upper_Bound)]
#Display Detected outlier
print("\nOutliers:")
Outliers
```

**Step 11 : count Outliers**

```python
# count Outliers
Outliers.shape[0]
```

**Step 12 : visualize Outliers using Box plot**

```python
#visualize Outliers using Box plot
plt.boxplot(df['Final_Marks'])
plt.ylabel('Final_Marks')
plt.title('Box Plot of Final_Marks')
plt.show()
```

**Step 13 :create Histogram**

```python
#Create Histogram To observe distribution of Final_marks
plt.hist(df['Final_Marks'], bins=10, edgecolor='black')
plt.xlabel('Final_Marks')
plt.ylabel('Frequency')
plt.title('Histogram of Final_Marks')
plt.show()
```

**Step 14 : Analyze Distribution Using Skewness**

```python
#Calculate skewnes of Final_Marks
Skewness = df['Final_Marks'].skew()
#Display Skewness
print("Skewness of Final_Marks:", Skewness)
```

**Step 15 : Calculate kurtosis**

```python
#Calculate kurtosis
Kurtosis = df['Final_Marks'].kurtosis()
#Display Kurtosis
print("Kurtosis of Final_Marks:", Kurtosis)
```

**Step 16 : Distribution of Study_Hours**

```python
#Create A Histogram for Study_Hours
plt.hist(df['Study_Hours'], bins=10, edgecolor='black')
plt.xlabel('Study_Hours')
plt.ylabel('Frequency')
plt.title('Histogram of Study_Hours')
plt.show()
```

**Step 17 : compare variable using box plot**

```python
#Create a box plots for multiple numerical variables
plt.boxplot([df['Study_Hours'],df['Attandance'] ,df['assignment_Score'], df['Final_Marks']])
plt.xticks([1, 2, 3,4], ['Study_Hours', 'Attandance','assignment_Score', 'Final_Marks'])
plt.ylabel('Value')
plt.title('Box Plots of Study_Hours,Attandance, assignment_Score, and Final_Marks')
plt.show()
```
