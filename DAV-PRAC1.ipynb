Step 1 : Import Pandas Library

```python
# Import the Pandas library for data manipulation and analysis
# The 'as pd' shortcut makes the code shorter and faster to type
import pandas as pd
```

Step 2 : Create DataSet

```python
# Create a dictionary containing sample student records (ID, name, age, city, and marks)
# This raw dictionary will be used next to build a structured data table
data = {
    "Id" :[101,102,103,104,105,106,107],
    "Name":["rahul","priya","amit","riya","karan", "krisha","harry"],
    "age":[20,21,19,20,22,24,28],
    "city":["surat","navsari","bharuch","gandhinagar","surat","valsad","ahemedabad"],
    "marks":[85,92,70,80,81,84,90]
}
```

Step 3 : Create DataFrame

```python
# Convert the dictionary into a Pandas DataFrame, which structures it like an Excel table
# Display the table to review the organized columns and rows of data
df =pd.DataFrame(data)
df
```

Step 4 : Save Data as CSV file

```python
# Save the data inside the dataframe (df) into a new file named "students.csv"
# index=False prevents pandas from adding an extra column of row numbers
df.to_csv("students.csv",index=False)
# Print a friendly message to let you know the file was successfully saved
print("csv file create successfully.")
```

Step 5 : Save Data as Excel file

```python
# Save the data inside the dataframe (df) into an Excel file named "students.xlsx"
# index=False prevents pandas from saving row numbers as a separate column
df.to_excel("students.xlsx",index=False)
# Print a message to confirm the file has been saved successfully
print("Excel file create successfully.")
```

Step 6 : Save Data as JSON file

```python
# Save the dataframe (df) as a JSON file named "students.json"
# orient="records" formats the data like a list of dictionaries (key-value pairs)
# indent=4 formats the file with nice spacing so it is clean and easy to read
df.to_json("students.json",orient="records",indent=4)
print("JSON file create successfully.")
# Print a message to confirm the file has been saved successfully
```
