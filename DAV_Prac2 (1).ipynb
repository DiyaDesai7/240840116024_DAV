####**AIM : Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.**

Step 1: Import pandas library

```python
## Import pandas so we can work with tables and data
import pandas as pd
```

Step 2 : Create DataSet

```python
# Store student names, ages, cities, and marks inside a dictionary
data = {
    "ID":[101,102,103,104,104,105,106,107],
    "Name": ["Rahul", "Priya", "Amit", None, "Riya", "Ajay", "Naresh", "Jainil"],
    "Age": [20, 21, None, 20, 20, 22, 23, 28],
    "City": ["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahmedabad"],
    "Marks": [85, None, 78, 88, 88, 81, 90, 93]
}
```

Step 3 : Create DataFrame

```python
# A DataFrame arranges the data into rows and columns like a table
df=pd.DataFrame(data)
df
```

step 4 : Display first 5 record

```python
# head() lets us quickly see the first five rows of our dataset
df.head()
```

step 5 : Display last 5 record

```python
# tail() is useful when we want to check the records at the bottom(last five rows)
df.tail()
```

step 6 : Display Information

```python
# info() tells us about columns, data types, and how many values are available
df.info()
```

step 7 : Display Statistical Summary

```python
# describe() gives useful values such as average, minimum, maximum, and count
df.describe()
```

Step 8 : Display Dimension of Dataset

```python
# shape tells us how many rows and columns are present
df.shape
```

Step 9 : Display Name of column

```python
# columns helps us know which fields are available in our dataset
df.columns
```

Step 10 : Display DataType of each column

```python
# dtypes tells us whether a column contains numbers, text, or decimal values
df.dtypes
```

Step 11 : Check missing values

```python
# isnull() finds empty cells and sum() counts them for each column df.isnull().sum()
df.isnull().sum()
```

Step 12 : Fill missing values in Marks column

```python
# Find the average marks and use it wherever the Marks value is missing
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

Step 13 : Fill missing values in Age column

```python
# Use the average age to replace the empty Age value
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

Step 14 : Fill missing values in Name column

```python
# Replace the empty name with "Unknown" so the cell is no longer blank
df["Name"] = df["Name"].fillna("unknown")
df
```

Step 15:Detect Duplicate records

```python
# duplicated() tells us which rows appear more than once
df.duplicated()
```

Step 16 : Count total no of Duplicate record

```python
# sum() adds all the True values to give us the total number of duplicates
df.duplicated().sum()
```

Step 17 : Remove Duplicate column

```python
# drop_duplicates() keeps one copy and removes repeated rows
df = df.drop_duplicates()
df
```

Step 18 :Rename column Marks to Score

```python
# rename() is used when we want to give a column a new name
df.rename(columns={"Marks": "Score"}, inplace=True)
df
```

Step 19 : Convert Age colum Datatype from float to int

```python
# astype(int) converts the Age values into whole numbers
df["Age"] = df["Age"].astype(int)
df.dtypes
```

Step 20 : Create new column Final score

```python
# Increase the Score by 10% and store the result in a new column
df["Final score"] = df["Score"] + 5
df
```

Step 21 : Convert all student name to uppercase

```python
# upper() converts names into Upper case
df["Name"] = df["Name"].str.upper()
df
```

Step 22 : Convert city name to lowercase

```python
#lower() converts city names such as "Surat" into "surat"
df["City"] = df["City"].str.lower()
df
```

Step 23 : Removw leading and trailing whitespace from City column

```python
#Remove extra spaces from the beginning and end of city names
df["City"] = df["City"].str.strip()
df
```

Step 24 : Sort data based on Final score in Descending order

```python
# Sort the data by Final Score from highest to lowest
df_sorted = df.sort_values(by="Final score", ascending=False)
# Display the sorted data
df_sorted
```

Step 25 : Save cleaned DataFrame as new CSV file.

```python
# Save the cleaned dataset as a CSV file
# index=False prevents pandas from adding an extra index column
df.to_csv("cleaned-students.csv", index=False)
# Show a message after the file is successfully created
print("Dataset Created Successfully")
```
