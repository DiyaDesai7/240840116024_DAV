###**AIM : Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in datasets.**

Step 1 : Import the Library

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
```

Step 2: create a dataset

```python
# Create a dataset with monthly sales and profit details
data = {
 "Month": ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September',
           'October', 'November', 'December'],
 "Units_sold":[120,135,150,145,160,175,180,170,190,210,230,250],
 "Revenue":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
 "Profit":[20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
}
```

```python
# Create a DataFrame from the dataset
df = pd.DataFrame(data)
# Display the dataset
df
```

Step 3 : Inspect Dataset

```python
# Display the first five rows
df.head()
```

```python
#Diplay no of row and column
df.shape
```

```python
# Display all column names
df.columns
```

```python
# Display information about dataset
df.info()
```

Step 4 : Generate Descriptive Statistic

```python
# Display descriptive statistics of numerical columns
df.describe()
```

Step 5 : Calculate Average number of units(mean)

```python
#display average no of units sold per month
average_units=df["Units_sold"].mean()
print("Average number of units sold per month:",average_units)
```

Step 6 : Find Median

```python
# Calculate the median units sold
median_units=df["Units_sold"].median()
# Display the median units sold
print("Median units sold:",median_units)
```

Step 7 : Find Maximum and Minimum

```python
#find highest units sold in month
maximum_units=df["Units_sold"].max()
print("Maximum unit sold:",maximum_units)
# Find the lowest number of units sold
minimum_units=df["Units_sold"].min()
print("Minimum unit sold:",minimum_units)
```

Step 8 : calculate standard deviation

```python
# Calculate the standard deviation of units sold
std_units=df["Units_sold"].std()
print("Standard deviation of units sold:",std_units)
```

Step 9 : Find Month with highest sales

```python
# Find the index of the month with the highest revenue
highest_sales_index=df["Revenue"].idxmax()

# Get the month with the highest revenue
highest_sales_month=df.loc[highest_sales_index,"Month"]

# Get the highest revenue amount
highest_revenue=df.loc[highest_sales_index,"Revenue"]

print("HIghest Sales month",highest_sales_month)
print("Highest Revenue:",highest_revenue)
```

Step 10 : Find Month with LOwest sales

```python
# Find the index of the month with the lowest revenue
lowest_sales_index=df["Revenue"].idxmin()

# Get the month with the lowest revenue
lowest_sales_month=df.loc[lowest_sales_index,"Month"]

# Get the lowest revenue amount
lowest_revenue=df.loc[lowest_sales_index,"Revenue"]
print("Lowest Sales month",lowest_sales_month)
print("Lowest Revenue:",lowest_revenue)
```

Step 11 : Total revenus

```python
# Calculate the total revenue
total_revenue=df["Revenue"].sum()
print("Total revenue:",total_revenue)
```

Step 12 : Calculate Total profit

```python
# Calculate the total profit
total_profit=df["Profit"].sum()
print("Total profit:",total_profit)
```

Step 13 : Identify Sales Trend

```python
# Create a line chart for monthly revenue
plt.figure(figsize=(10,5))
# Plot revenue for each month
plt.plot(df["Month"],df["Revenue"],marker='o')
# Add labels and title
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Monthly Sales Revenue trend")
# Rotate month names for better visibility
plt.xticks(rotation=45)
# Display the chart
plt.show()
```

Step 14 : create line chart for monthly profit

```python
from matplotlib.lines import lineStyles
# Create a line chart for monthly revenue
plt.figure(figsize=(10,5))
# Plot revenue for each month
plt.plot(df["Month"],df["Profit"],marker='o',linestyle='--')
# Add labels and title
plt.xlabel("Month")
plt.ylabel("Profit")
plt.title("Monthly Profit trend")
# Rotate month names for better visibility
plt.xticks(rotation=45)
# Display the chart
plt.show()
```

Step 15 : compare Monthly revenue and Profit

```python
# Create a chart to compare revenue and profit
plt.figure(figsize=(10,5))
# Plot monthly revenue
plt.plot(df["Month"],df["Revenue"],marker='o',label="Revenue")
# Plot monthly profit
plt.plot(df["Month"],df["Profit"],marker='o',label="Profit")
# Add labels and title
plt.title("Revenue vs Sales")
plt.xlabel("Month")
plt.ylabel("Amount")
# Display the chart legend
plt.legend()
# Rotate month names for better visibility
plt.xticks(rotation=45)
# Display the chart
plt.show()
```

Step 16 : Identify Growth in Sales

```python
# Calculate the percentage growth in revenue
df["Revenue_Growth_%"]=df['Revenue'].pct_change()*100
df
```
