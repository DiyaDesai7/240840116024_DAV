###**AIM : Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).**

**Step 1 : Import Require Library**

```python
#import requiered libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**Step 2 : Create DataFrame for Given Dataset**

```python
#create Dataframe
data = {
    "Month" : ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Product" : ["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet",
                 "Laptop","Mobile","Tablet"],
    "Region" : ["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales" : [120000,135000,150000,146000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_sold" : [12,30,25,14,35,30,18,38,32,21,45,40]
}
df = pd.DataFrame(data)
#Display DataFrame
display(df)
```

**Step 3 : Create a line chart**

```python
#Create a line chart for Month wise Sales
plt.plot(df["Month"],df["Sales"],marker = 'o')
plt.title("Sales over the month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

**Interpretation of Line Chart:**


*   The line chart shows the monthly sales trend from January to December.

  * Sales started at ₹1,20,000 in January and increased to ₹2,50,000 in December.

* Sales generally show an increasing trend throughout the year.

* There was a small decrease in sales from March (₹1,50,000) to April (₹1,46,000).

* There was another decrease from July (₹1,80,000) to August (₹1,70,000).

* December recorded the highest sales of ₹2,50,000.

* January recorded the lowest sales of ₹1,20,000.

 * Overall, the chart indicates that sales improved significantly during the year




**Step 4 : Create a BarChart**

```python
#create a barchart product vs sales
plt.bar(df["Product"],df["Sales"],edgecolor='black')
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

**Interpretation of Bar Chart:**
* The bar chart compares sales for different products across the months.

* Tablet sales increased from ₹1,50,000 in March to ₹2,50,000 in December.

* Mobile sales increased from ₹1,35,000 in February to ₹2,30,000 in November.

* Laptop sales increased from ₹1,20,000 in January to ₹2,10,000 in October.

* The highest individual sales value is ₹2,50,000 for Tablet in December.

* The lowest individual sales value is ₹1,20,000 for Laptop in January.

* Overall, the chart shows that sales values increased for all three products over their respective months.

**Step 5 : Create a Histogram**

```python
#create Histogram on how individual sales values distributed
plt.hist(df["Sales"],bins=5,edgecolor='black')
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

**Interpretation of Histogram :**

* The histogram shows the distribution of monthly sales values.

* The sales values range from ₹1,20,000 to ₹2,50,000.

* Most of the sales values are concentrated in the middle-to-higher sales range.

* There are fewer observations at the lower end, such as ₹1,20,000 and ₹1,35,000.

* Higher sales values such as ₹2,10,000, ₹2,30,000 and ₹2,50,000 occur toward the end of the year.

* The histogram indicates that the dataset has a general shift toward higher sales values.

**Step 6 : create a Scatter plot**

```python
#create a scatter plot for relationship between units sold and sales
plt.scatter(df["Units_sold"],df["Sales"])
plt.title("Relationship between Units Sold and Sales")
plt.xlabel("Units Sold")
plt.ylabel("Sales")
plt.show()
```

**Interpretation of Scatter Plot:**
* The scatter plot shows the relationship between Units Sold and Sales.

* There is a clear positive relationship between units sold and sales.

* In general, when the number of units sold increases, sales also increase.

* For example, 12 units sold corresponds to ₹1,20,000 sales, while 45 units sold corresponds to ₹2,30,000 sales.

* The highest number of units sold is 45 in November, with sales of ₹2,30,000.

* The lowest number of units sold is 12 in January, with sales of ₹1,20,000.

* Therefore, the scatter plot indicates that higher unit sales are generally associated with higher revenue.

**Step 7 : Create a Pie chart**

```python
#crate a pie chart what percentage of total sales comes from each region
region_sales = df.groupby("Region")["Sales"].sum()
plt.pie(region_sales,labels=region_sales.index,autopct='%1.1f%%')
plt.title("Distribution of Sales by Region")
plt.show()
```

**Interpretation of Pie Chart:**
* The pie chart shows the percentage contribution of each region to total sales.

* The total sales are ₹20,16,000.

* North has the highest total sales of ₹5,10,000, contributing approximately 25.3%.

* East has total sales of ₹5,80,000, contributing approximately 28.8%.

* South has total sales of ₹5,40,000, contributing approximately 26.8%.

* West has total sales of ₹3,86,000, contributing approximately 19.1%.

* Therefore, East is the highest contributing region, while West is the lowest contributing region.

* The chart shows that the sales contribution is not equal across all regions.
